### dateTime column main convert karna zaroori hai so that pata chal sake k customer kitne time sy hamary sath hai

```python
data['customer_since'] = pd.to_datetime(data['customer_since'])

# ispr time series analysis kr k dekhte lete hain
data['customer_tenure_days'] = (pd.to_datetime('today') - data['customer_since']).dt.days
```

```python
data = data.drop(columns=['customer_since'])
```

```python
data.head(2)
```

```python
data.duplicated(subset=['customer_id']).sum()
```

### Un_use_full columns ko drop is lea kr raha hun k data sy noise kam ho jae

```python
data.drop(columns=['customer_id']).head(5)
```

### Kuch Missing Values hain unko theek kr raha hun
- avg_order_value (250) float type
- email_open_rate (250) float type

```python
data['avg_order_value'] = data['avg_order_value'].fillna(data['avg_order_value'].median())
```

```python
data['email_open_rate'] = data['email_open_rate'].fillna(data['email_open_rate'].median())
```
