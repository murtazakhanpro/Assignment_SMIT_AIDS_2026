# Step 3: Feature Engineering
        Raw total features (eg. Total rooms ya total population)
        - rooms_per_household: Average rooms per house.
        - bedrooms_per_room: Propotional of bedrooms relative to                total rooms.
        - population_per_household: Density of occupants per house.

```python
# direct room ratios calculate kar k dekh leta hain
df['rooms_per_household'] = df['total_rooms'] / df['households']
df['bedroom_per_room'] = df['total_bedrooms'] / df["total_rooms"]
df['population_per_household'] = df['population'] / df['households']
```

# Step : 4 Train-Test Split & Feature Selection
<br>Explanation<br>
<p>Target column median_house_value ko features X se  seprate karain gy. preprocessing ke liye numerical aur categorail columns ko define karne ke baad dataset ko 80% Training aur 20% Testing sets me split karte hain. random_state=42 reproducible results ke liye set karte hain.</p>

```python
# usaman bhai why do i forgot the repeated process like, test, train, split,
```

```python
#feature (X) aur Target (y) split
X = df.drop('median_house_value', axis=1)
y = df['median_house_value']
```

```python
# Categorial vs Numerical Feature identification
num_cols = X.select_dtypes(include=['int64', 'float64']).columns.tolist()
cat_cols = ['ocean_proximity']
```

```python
# Train-Test Split (80# Train, 20% Test)
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.20, random_state=42)
```

```python
print(f"X_train shape: {X_train.shape} | X_test shape: {X_test.shape}")
```

# Step 5: Data Preprocessing Pipeline Construction
Explation:
Data leake avoid karne aur modular trasformaiton k liye Scikit_learn pipelines use karta hai.

1: numercial Pipeline: SimpleImputer (media strategy) se missing values fill karte hain aur (standardScaler) se numerical features ko standard normal distribution (mean=0, std=1) me bring krte hain.

2: Categorcal Pipeline: Categorical variable occean_prximity ko OneHotEncoder(drop='first') se encode krte hain taake dummy variable trap se bacha ja sake

3: ColumnTransformer: Dono pipelines ko single interface me join krta ha.

```python
# PipeLine for Numerical Features
num_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])
```

```python
# Pipeline for categorical Features
cat_pipeline = Pipeline([
    ('encoer', OneHotEncoder(drop='first', handle_unknown='ignore'))
])
```

```python
# Master ColumnTransformer
preprocessor = ColumnTransformer([
    ('num', num_pipeline, num_cols),
    ('cat', cat_pipeline, cat_cols)
])
```

# Step 6: Model Training & Comparative Evaluation
Explaination
- Multiple Regression algorithms(LinearRegression, DecisionTreeRegressor, RandomForestRegressor, GredientBostingRegressor)
- metrices(MAE, MSE, RMSE, R2 Score) par evaluate aur compare krte hain

```python
models = {
    'Linear Regression': LinearRegression(),
    'Decision Tree': DecisionTreeRegressor(random_state=42),
    'Random Forest': RandomForestRegressor(random_state=42, n_jobs=-1),
    'Gradient Boosting': GradientBoostingRegressor(random_state=42)
}

results = []

for name, model in models.items():
    # Model Pipeline combine karein
    full_pipeline = Pipeline([
        ('preprocessor', preprocessor),
        ('model', model)
    ])
    
    # Train
    full_pipeline.fit(X_train, y_train)
    
    # Predict on Test Set
    y_pred = full_pipeline.predict(X_test)
    
    # Calculate Metrics
    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, y_pred)
    
    results.append({
        'Model': name,
        'MAE': round(mae, 2),
        'MSE': round(mse, 2),
        'RMSE': round(rmse, 2),
        'R2 Score': round(r2, 4)
    })

# Metrics comparison table
comparison_df = pd.DataFrame(results).sort_values(by='R2 Score', ascending=False)
comparison_df
```

# Step 7: Hyperparameter Tuning
(Best Model Optimization)Explanation:Random Forest Regression sab se superior R2 score ($> 0.81$) offer karta hai. 
Is broad performance ko further refine karne ke liye GridSearchCV se hyperparameter optimization apply karte hain.  

```python
# Tuning pipeline setup
tuning_pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('model', RandomForestRegressor(random_state=42, n_jobs=-1))
])

# Search grid define karein
param_grid = {
    'model__n_estimators': [100, 200],
    'model__max_depth': [20, None],
    'model__max_features': ['sqrt', 1.0]
}

grid_search = GridSearchCV(
    estimator=tuning_pipeline,
    param_grid=param_grid,
    cv=3,
    scoring='r2',
    n_jobs=-1
)

grid_search.fit(X_train, y_train)

print("Best Parameters:", grid_search.best_params_)
print("Best CV R2 Score:", round(grid_search.best_score_, 4))
```
